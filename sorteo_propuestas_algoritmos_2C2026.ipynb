```python
import pandas as pd
import numpy as np
```

```python
#Importo el excel para poder trabajar
df= pd.read_excel("presentismo_primer_parcial_2C2026_algoritmos_xlsx")
df
```

```python
# 1. Limpiar espacios en blanco
df["Ingrese su Apellido"] = df["Ingrese su Apellido"].str.strip()
df["Ingrese su Nombre"] = df["Ingrese su Nombre"].str.strip()

# 2. Eliminar duplicados por Nombre y Apellido (o por correo)
# keep='last' se queda con la última respuesta que envió el alumno
df = df.drop_duplicates(subset=["Ingrese su Apellido", "Ingrese su Nombre"], keep="last")

# 3. Ordenar alfabéticamente el DataFrame completo y reiniciar índices
df = df.sort_values(by=["Ingrese su Apellido", "Ingrese su Nombre"]).reset_index(drop=True)
```

```python
#Me quedo unicamente con la columna "Apellido"
lista_apellidos=df["Ingrese su Apellido"]
lista_nombres=df["Ingrese su Nombre"]
temas_parcial= ["Propuesta 1", "Propuesta 2", "Propuesta 3", "Propuesta 4", "Propuesta 5"]

print(lista_apellidos)
print(temas_parcial)
```

```python
# Cantidad total de alumnos y temas
cant_alumnos = len(lista_apellidos)
cant_temas = len(temas_parcial)
```

```python
# 1. Armo la lista perfectamente balanceada
asignacion_temas = (temas_parcial * (cant_alumnos // cant_temas + 1))[:cant_alumnos]

# 2. Mezclar los temas de forma aleatoria
rng = np.random.default_rng()
rng.shuffle(asignacion_temas)

# 3. Creo el DataFrame final combinando la lista de alumnos y los temas desordenados
df_sorteo = pd.DataFrame({
    "Apellido": lista_apellidos,
    "Nombre": lista_nombres,
    "Tema": asignacion_temas
})
df_sorteo = df_sorteo.sort_values(by=["Apellido", "Nombre"]).reset_index(drop=True)
df_sorteo
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, len(df_sorteo) * 0.4 + 1))
ax.axis("off")
ax.axis("tight")

tabla = ax.table(
    cellText=df_sorteo.values,
    colLabels=df_sorteo.columns,
    cellLoc="center",
    loc="center"
)
tabla.auto_set_font_size(False)
tabla.set_fontsize(10)
tabla.scale(1.2, 1.2)

plt.savefig("primer_parcial_sorteo_temas_algoritmos_2C2026.png", bbox_inches="tight", dpi=300)
plt.close()
```
