## 3. Tiempo de encendido vs. puntaje de experiencia de uso de máquina
**Relación:** ¿A mayor tiempo de encendido, menor puntaje?

Cuando trabajamos con dos columnas numéricas podemos calcular el **coeficiente de
correlación de Pearson**. Se propone investigar sobre el mismo para poder usarlo en el análisis. Como introducción breve, el coeficiente de correlación es un número entre -1 y 1 que mide cómo están relacionadas dos variables *de cierta forma* y con qué fuerza:

    - Cerca de 1: relación directa — cuando una sube, la otra sube.
    - Cerca de -1: relación inversa — cuando una sube, la otra baja.
    - Cerca de 0: no hay relación.
    - Mientras más cerca de 1 o -1, más fuerte es la relación.

    Ojo: correlación no implica causalidad. Mide si las variables se mueven juntas,
    no dice que una sea la causa de la otra.
